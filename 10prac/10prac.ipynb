Задание 1  
Анализ логов: Только ошибки  
◆ Условие  
Откройте файл server_log.txt для чтения. Прочитайте все строки. Создайте новый файл errors_only.txt. Запишите в него только строки,   содержащие слово ERROR или CRITICAL. Выведите в консоль сообщение: «Найдено X критических ошибок».  

```python
cnt = 0
with open("./files/server_log.txt", "r", encoding="utf-8") as f:
    with open("./my_files/errors_only.txt", 'w', encoding="utf-8") as nf:
        for line in f:
            if 'ERROR' in line or 'CRITICAL' in line:
                nf.write(line)
                cnt += 1

print(f"Найдено {cnt} критических ошибок")
```
