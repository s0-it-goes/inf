```python
cnt = 0
with open("./files/server_log.txt", "r", encoding="utf-8") as f:
    with open("./my_files/errors_only.txt", 'w', encoding="utf-8") as nf:
        for line in f:
            if 'ERROR' in line or 'CRITICAL' in line:
                nf.write(line)
                cnt += 1

print(f"Найдено {cnt} критических ошибок")
```

Задание 4

```python
with open("./files/secret.txt", 'r', ) as f:
    with open("./my_files/secret_anonymized.txt", 'w', encoding='utf-8') as mf:
        # заношу в список каждую строку файла, предварительно убрав из нее \n, слова тел: перед телефоном и email: перед email'ом
        # потом метод split разделяет каждый элемент по запятой и возвращает список
        l = [line.rstrip('\n').replace('тел: ', '').replace('email: ', '').split(', ') for line in f]

        for line in l:
            # заменяю все символы, кроме последних двух с помощью replace. количество звездочек считаю через длину среза
            line[1] = ', ' + line[1].replace(line[1][:-2], '*' * len(line[1][:-2])) + ', '
            i = line[2].find('@') # индекс @

            # заменяю срез до @ звездочками
            line[2] = line[2].replace(line[2][1:i], '*' * len(line[2][1:i]))
            #print(line)
            mf.write(''.join(line) + '\n')
```

ЗАДАНИЕ 5

```python
with open("./files/article.txt", 'r', encoding='utf-8') as f:
    words = (f.read()).lower().split(' ')

    words.remove('')
    
    d = {}

    for i in range(len(words)):
        if words[i] and not (words[i][-1].isalpha()):
            words[i] = words[i][:-1]

        # != None, т.к если words[i] не находится в словаре, то возвращает None
        d[words[i]] = d[words[i]] + 1 if d.get(words[i]) != None else 1

    d = sorted(d.items(), key=lambda x: x[1], reverse=True)

    for i in range(5):
        print(d[i])
    
```

```python
d = {
    '2': 'ABC',
    '3': 'DEF',
    '4': 'GHI',
    '5': 'JKL',
    '6': 'MNO',
    '7': 'PQRS',
    '8': 'TUV',
    '9': 'WXYZ',
    '0': ' '
}

def deshifrate(st):
    st = st.upper()
    newstr = ''

    for i in st:
        if i.isalpha() or i == ' ':
            newstr += i

    newstr = list(newstr)
    
    items = d.items()
    
    for num, letters in items:
        for i in range(len(newstr)):
            if newstr[i] in letters:
                newstr[i] =  num * (letters.find(newstr[i]) + 1)
    
    print(' '.join(newstr))

deshifrate('DFfs??gfg 5g')
```

```python
import json

with open("./files/servers.json", 'r', encoding='utf-8') as f:
    d = json.load(f)
    online = offline = maintenance = 0
    for name, value in d.items():
        if value == 'online':
            online += 1
        elif value == 'offline':
            offline += 1
            print(name)
        else:
            maintenance += 1

    print(f"\nStatistics:\nonline: {online}\noffline: {offline}\nmaintenance: {maintenance}")
```

ЗАДАНИЕ 8

```python
import json

with open("./files/playlist.json", 'r', encoding='utf-8') as f:
    d = json.load(f)
    styles = []
    for style in d:
        styles.append(style)
        print(style)

    print("\nКакой жанр хотите послушать?")
    us_input = input()

    while(us_input not in styles):
        us_input = input()

    print(*d[us_input], sep='\n')

    print("\nКакой трек добавляем? Назавание трека и жанр")
    name = input()
    style = input()

    d[style].append(name)

with open("./files/playlist.json", 'w', encoding='utf-8') as f:
    json.dump(d, f, ensure_ascii=False, indent=4)
```
