### Задание 1: Разворот цифр числа

```python
def reverse_number(n):
    rev = 0

    while(n > 0):
        digit = n % 10
        rev = rev * 10 + digit
        n //= 10
    
    return rev

print(reverse_number(1234))
```

### Задание 2: Наибольший общий делитель (алгоритм Евклида)

```python
def gcd(a, b):
    while b != 0:
        temp = b
        b = a % b
        a = temp

    return a
```

### Задание 3: Проверка числа на простоту

```python
def isprime(n):
    if n <= 1:
        return False

    d = 2

    while d**2 <= n:
        if n % d == 0:
            return False

        d += 1

    return True
```

### Задание 4: N-е число Фибоначчи

```python
def fibonacci(n):
    if n == 0:
        return 0

    prev = 0
    curr = 1
    i = 1

    while i < n:
        nxt = prev + curr
        prev = curr
        curr = nxt
        i += 1

    return curr
```

### Задание 5: Наибольшая последовательность возрастающих элементов

```python
def max_increasing_len(arr):
    if len(arr) == 0:
        return 0

    cur = best = i = 1

    while i < len(arr):
        cur = cur + 1 if arr[i] > arr[i-1] else 1

        best = max(best, cur)

        i += 1

    return best
```

### Задание 6: Квадратный корень методом Ньютона (Герона)

```python
def sqrt_newton(n, eps=1e-6):
    if n == 0:
        return 0.0

    x = n / 2

    while abs(x**2 - n) > eps:
        x = 0.5 * (x + n / x)

    return x

```

### Задание 7: Поиск второго максимума

```python
def second_max(arr):
    if len(arr) < 2:
        return None

    m1 = m2 = float("-inf")
    i = 0

    while i < len(arr):
        if arr[i] > m1:
            m2 = m1
            m1 = arr[i]
        elif m2 < arr[i] and arr[i] < m1:
            m2 = arr[i]

        i += 1

    return m2

```

### Задание 8: Количество простых чисел в диапазоне от 2 до N

```python
def count_primes_up_to(n):
    cnt = 0
    num = 2

    while num <= n:
        is_p = True
        d = 2

        while d**2 <= num and is_p:
            if num % d == 0:
                is_p = False

            d += 1

        if is_p:
            cnt += 1

        num += 1

    return cnt

        
```
