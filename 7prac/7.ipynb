```python
students = [
    ("Аня", 5), ("Борис", 3), ("Вера", 4),
    ("Жанна", 5), ("Дина", 2), ("Егор", 4),
    ("Глеб", 5), ("Зоя", 3)
]

students = filter(lambda x: x[1] >= 4, students)
students = sorted(students, key=lambda x: (x[1], x[0]))

mean = sum(s[1] for s in students) / len(students)

print(*students, f"avg: {mean}", sep = '\n')
```
