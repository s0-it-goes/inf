```python
import random

n :int = int(input())
treasure :list = [[random.randint(1, 10), random.randint(1, 10)] for i in range(n)]
alex :list = [int(input()), int(input())]

print(f"Количество сокровищ: {n}", f"Координаты:", *treasure, f"Координаты Александра: {alex}", sep='\n')

nearest_treasure = treasure[0]
l = ((alex[0] - nearest_treasure[0])**2 + (alex[1] - nearest_treasure[1])**2)**0.5

for t in treasure:
    tmp_length = ((alex[0] - t[0])**2 + (alex[1] - t[1])**2)**0.5
    if tmp_length < l:
        l = tmp_length
        nearest_treasure = t

print(f"Ближайшее Сокровище: {nearest_treasure}", f"Расстояние: {l}", sep = '\n')
    
```
