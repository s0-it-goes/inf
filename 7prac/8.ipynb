```python
import random
import sys

n, m, k = (int(input()) for _ in range(3))


matrix = [[random.randint(0, 1) for i in range(n)] for j in range(m)]

print(*matrix, sep = '\n')

if k > m:
    print("Количество зрителей больше ширины ряда")
    sys.exit(0)

cnt = 0

for i in range(n):
    for j in range(m):
        cnt = cnt + 1 if matrix[i][j] == 0 else 0
        if cnt >= k:
            break

    if cnt >= k:
        print(i+1)
        break
    cnt = 0

```
