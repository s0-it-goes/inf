```python
def tokenize(text: str):
    separators = '.=,;:!?()[]{}"«»…—–\'- '

    tokens = []
    tmp = ''

    for i in range(len(text) - 1):
        if text[i] not in separators:
            tmp += text[i]
            continue
        else:
            if (text[i] == "\'" or text[i] == "-") and (text[i+1].isalpha()) and (tmp.isalpha()) :
                tmp += text[i]
                continue
            elif (text[i] == '.' and text[i+1].isdigit() and tmp.isdigit()):
                tmp += text[i]
                continue
            if tmp:
                tokens.append(tmp)

            tmp = ''

    if text and text[-1] not in separators:
        tmp += text[-1]
    tokens.append(tmp)

    print(tokens)

tokenize("-Привет, мир! Как дела?")    # ["Привет", "мир", "Как", "дела"]
tokenize(".3 Жили-были 3.14 раза")        # ["Жили-были", "3.14", "раза"]
tokenize("s = \"don't\"")              # ["s", "don't"]
tokenize("A...B")                      # ["A", "B"]
tokenize("")                           # []
```
