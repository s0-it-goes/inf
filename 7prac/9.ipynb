```python
n, m = (int(input()) for _ in range(2)) 

A = [[0 for i in range(m)] for j in range(n)]

for i in range(m):
    A[0][i] = 1

for j in range(n):
    A[j][0] = 1

for i in range(1, n):
    for j in range(1, m):
        A[i][j] = A[i-1][j] + A[i][j-1]

for i in A:
    for x in i:
        print(f"{x:6}", end = " ")
    print()
```
