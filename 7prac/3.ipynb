```python
import random

n :int = int(input())

def make_magic_square(N: int):

    square :list = [[0] * N for i in range(N)]

    position :list = [0, N // 2]
    cnt :int = 0

    number :int = 0

    for i in range(N):
        for j in range(N):
            number += 1

            if cnt == N:
                position[0] += 2
                position[1] -= 1
                cnt = 0
            if position[1] >= N:
                position[1] = 0
            if position[0] < 0:
                position[0] = N-1

            square[position[0]][position[1]] = number

            position[0] -= 1
            position[1] += 1
            
            cnt += 1       

    return square

def make_random_magic_square(N: int , square: list):
    for i in range(random.randint(0, 3)): # поворачиваем квадрат на 90 градусов против часовой стрелки
        square = [list(row) for row in list(zip(*square))[::-1]]

    if random.choice([True, False]): # отражаем квадрат по вертикали
        square = square[::-1]

    if random.choice([True, False]): # отражаем квадрат по горизонтали
        square = [row[::-1] for row in square]

    if random.choice([True, False]): # меняем каждый элемент квадрата на N**2 + 1 - x
        square = [[N**2 + 1 - x for x in row] for row in square]
    
    return square

magic_constant = n * (n**2 + 1) / 2

def count_magic_constant(N: int, square: list, constant: int):
    sums :int = [sum(row) for row in square]
    sums += [sum(col) for col in list(zip(*square))]
    sums.append(sum(square[i][i] for i in range(N)))
    sums.append(sum(square[i][N-1-i] for i in range(N)))

    return all(s == constant for s in sums)

x :list = make_magic_square(n)

for i in range(n):
    print(x[i])

print()

x = make_random_magic_square(n, x)

for i in range(n):
    print(x[i])

print(count_magic_constant(n, x, magic_constant))
```
