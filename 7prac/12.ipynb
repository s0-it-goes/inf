```python
md = input()

def markdown_to_html(md):
    cnt = 0

    
        
    if md and md[0] != '#':
        md = "<p>" + md + "</p>"
    else:
        for i in range(len(md)):
                if md[i] == '#':
                    cnt += 1
                else:
                    if cnt:
                        if md[i] == ' ':
                            md = f"<h{cnt}>" + md[i+1:]
                        else:
                            md = f"<h{cnt}>" + md[i:]
        
                        md += f"</h{cnt}>"
                        break
                    cnt = 0

    if md.count("**") != 0 and md.count("**") % 2 == 0 :
        for i in range(md.count("**")):
            if i % 2 == 0:
                md = md.replace("**", "<b>", 1)
            else:
                md = md.replace("**", "</b>", 1)

    if md.count("`") != 0 and md.count("`") % 2 == 0 :
            for i in range(md.count("`")):
                if i % 2 == 0:
                    md = md.replace("`", "<code>", 1)
                else:
                    md = md.replace("`", "</code>", 1)

    return md

print(markdown_to_html(md))

assert markdown_to_html("# Привет") == "<h1>Привет</h1>"
assert markdown_to_html("### Тема") == "<h3>Тема</h3>"
assert markdown_to_html("###### Мелкий") == "<h6>Мелкий</h6>"
assert markdown_to_html("Это **важно**") == "<p>Это <b>важно</b></p>"
assert markdown_to_html("Код `x`") == "<p>Код <code>x</code></p>"
assert markdown_to_html("") == ""
```
